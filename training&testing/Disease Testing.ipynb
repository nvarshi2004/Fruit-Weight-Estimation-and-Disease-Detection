```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.applications.efficientnet import preprocess_input

# ================= CONFIG =================

MODEL_PATH = r"F:\dd\archive\Research_Results\EfficientNet_Merged\EfficientNet_Final.keras"
IMAGE_PATH = r"F:\Testing apples\fruit_652_2.jpg"
IMG_SIZE = 224

# ================= LOAD MODEL =================

model = load_model(MODEL_PATH)

class_names = ['Blotch_Apple', 'Normal_Apple', 'Rot_Apple', 'Scab_Apple']

# ================= GRADE LOGIC =================

grading_system = {
    'Normal_Apple': {"grade": "Grade 1", "status": "Healthy"},
    'Blotch_Apple': {"grade": "Grade 2", "status": "Mild Disease"},
    'Scab_Apple': {"grade": "Grade 2", "status": "Moderate Disease"},
    'Rot_Apple': {"grade": "Grade 3", "status": "Severe Rot"}
}

# ================= PREPROCESS IMAGE =================

img = load_img(IMAGE_PATH, target_size=(IMG_SIZE, IMG_SIZE))
img_array = img_to_array(img)
img_array = preprocess_input(img_array)
img_array = np.expand_dims(img_array, axis=0)

# ================= PREDICT =================

prediction = model(img_array, training=False).numpy()
predicted_index = np.argmax(prediction)
predicted_class = class_names[predicted_index]

grade_info = grading_system[predicted_class]

# ================= OUTPUT =================

print("\n==============================")
print("Disease Detected :", predicted_class.replace("_Apple",""))
print("Grade            :", grade_info["grade"])
print("Health Status    :", grade_info["status"])
print("==============================")

# Show image
```
